```python
# file_loc = 'E:/SKRIPSI/code/Test.docx'
# file_loc = 'E:/SKRIPSI/Proposal Skripsi_Hugo Alfedo Putra_225150201111013.docx'
file_loc = "F:/DiskE/SKRIPSI/Skripsi_Hugo Alfedo Putra_225150201111013_CAMBRIA.docx"

doc = docx.Document(file_loc)

all_paras = doc.paragraphs
len(all_paras)
```

```python
val_dic_loc = 'C:/Users/guido/Downloads/full_validation_dictionary.txt' 
indo_dic_loc = 'F:/DiskE/SKRIPSI-CODE/skripsi-code/prep/kbbi_words.txt'
indo_dic = []
val_dic = []

file = open(val_dic_loc, 'r')
raw_val_dic = file.readlines()

for raw_item in raw_val_dic:
    val_dic.append(raw_item.strip())

print('engval dict size', len(val_dic))
file.close()

file = open(indo_dic_loc, 'r', encoding='utf-8')
raw_indo_dic = file.readlines()

for raw_item in raw_indo_dic:
    indo_dic.append(raw_item.strip())

print('indo dict size', len(indo_dic))
file.close()
```

```python
okay_list = [
    'CNN',
    'GRU',
    'LSTM',
    'DFT',
    'FFT',
    'STFT',
    'FC',
    'MER',
    'MIR',
    'CV',
    'MediaEval',
    'ROC',
]
```

```python
import string

translator = str.maketrans('', '', string.punctuation)

bib_texts = []
start_reading = 0
for para in all_paras:
    if para.text == "DAFTAR REFERENSI":
        start_reading = 1
        continue
    
    if start_reading == 0:
        continue
    else:
        runs = para.runs
        for run in runs:
            text = run.text.translate(translator)
            tokens = text.split(' ')

            if len(tokens) <= 1:
                continue
            # print(tokens)

            for token in tokens:
                # print(token)
                if len(token) > 1 and token[0].isupper():
                    bib_texts.append(token)
                if token.startswith('20'):
                    # print('ANOMALY FOUND')
                    break

print(len(bib_texts))
# print(all_authors)
```

```python
from pprint import pprint

translator = str.maketrans('', '', string.punctuation)

found_count = 0
found_list = set()

for para in all_paras:
    if para.text == "DAFTAR REFERENSI":
        break

    runs = para.runs
    for run in runs:
        text = run.text.translate(translator)

        isTextItalic = False if run.italic is None else True
        # It is now assumed that all the tokens in the text were not italicised
        tokens = text.split(' ')
        for token in tokens:
            if token.upper() in val_dic and token not in okay_list and token not in bib_texts and token.lower() not in indo_dic and not isTextItalic:
                # means that there is an English word that was not italicised (naive check)
                token_copy = tokens.copy()
                result = []
                for ctoken in token_copy:
                    if ctoken == token:
                        result.append(f'<<<{token.upper()}>>>')
                    else:
                        result.append(ctoken)
                print(f'Missing italics at word <<<{token}>>> in:\n{' '.join(result)}\n')
                found_count += 1
                found_list.add(token.lower())

if found_count > 0:
    print('There are missing italics!\n')
else:
    print('All done!')

pprint(found_list)
```
