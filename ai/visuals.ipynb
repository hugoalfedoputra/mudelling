```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n_points = 200

# Target Pearson corrs for visualisation sake
target_r_pos = 0.83
target_r_neg = -0.84
target_r_none = 0.0

mean = [0, 0]

# Covariance matrices [variance_x, covariance], [covariance, variance_y]
# Since standard deviations are 1, the covariance equals the Pearson r
cov_pos = [[1, target_r_pos], [target_r_pos, 1]]
cov_neg = [[1, target_r_neg], [target_r_neg, 1]]
cov_none = [[1, target_r_none], [target_r_none, 1]]

# Generate the random data based on the covariance matrices
x_pos, y_pos = np.random.multivariate_normal(mean, cov_pos, n_points).T
x_neg, y_neg = np.random.multivariate_normal(mean, cov_neg, n_points).T
x_none, y_none = np.random.multivariate_normal(mean, cov_none, n_points).T

# Calculate the actual Pearson r of the generated samples to display in the legend
# np.corrcoef returns a matrix, [0, 1] extracts the r value between x and y
actual_r_pos = np.corrcoef(x_pos, y_pos)[0, 1]
actual_r_neg = np.corrcoef(x_neg, y_neg)[0, 1]
actual_r_none = np.corrcoef(x_none, y_none)[0, 1]

# Create a figure with 1 row and 3 columns of subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Subplot 1: Positive Correlation
axes[0].scatter(x_pos, y_pos, label=f'Pearson r = {actual_r_pos:.4f}')
axes[0].set_title('Korelasi Positif')
axes[0].legend()
axes[0].grid()

# Subplot 2: Negative Correlation
axes[1].scatter(x_neg, y_neg, label=f'Pearson r = {actual_r_neg:.4f}')
axes[1].set_title('Korelasi Negatif')
axes[1].legend()
axes[1].grid()

# Subplot 3: No Correlation
axes[2].scatter(x_none, y_none, label=f'Pearson r = {actual_r_none:.4f}')
axes[2].set_title('Tidak Ada Korelasi')
axes[2].legend()
axes[2].grid()

plt.tight_layout()
plt.show()
```
